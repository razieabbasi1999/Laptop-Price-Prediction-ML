# **Preprocessing the " ScreenResolution " column :**


```python
import numpy as np
import re
```

```python
# ۱: IPS  ,  Touchscreen
df['IPS'] = df['ScreenResolution'].apply(lambda x: 1 if 'IPS' in x else 0)
df['Touchscreen'] = df['ScreenResolution'].apply(
lambda x: 1 if 'Touchscreen' in x else 0)
```

```python
# 2.1: Height and width extraction function
def extract_resolution(res):
  match = re.search(r'(\d+)x(\d+)', res)
  if match:
    return int(match.group(1)), int(match.group(2))
  return None, None


# 2.2: Creating columns X_res and Y_res
res_tuples = df['ScreenResolution'].apply(extract_resolution)
df['X_res'], df['Y_res'] = zip(*res_tuples)
```

```python
# 4: PPI calculation
df['PPI'] = (
(df['X_res'] ** 2 + df['Y_res'] ** 2) ** 0.5 / df['Inches']).round(2)
```

# **Preprocessing the " RAM " column :**

```python
df['Ram'] = df['Ram'].str.replace('GB', '').astype('int64')
```

# **Preprocessing the " Weight " column :**

```python
df['Weight'] = df['Weight'].str.replace('kg', '').astype('float64')
```

# **Preprocessing the " Memory " column :**

```python
def extract_all_memory_features(mem):
    mem = str(mem)

    # 1. SSD
    has_ssd = 1 if 'SSD' in mem else 0
    ssd_size = 0
    if has_ssd:
        ssd_gb = re.findall(r'(\d+)GB\s*SSD', mem)
        ssd_size += sum([int(x) for x in ssd_gb])
        ssd_tb = re.findall(r'(\d+)TB\s*SSD', mem)
        ssd_size += sum([int(x) * 1024 for x in ssd_tb])

    # 2. HDD
    has_hdd = 1 if 'HDD' in mem else 0
    hdd_size = 0
    if has_hdd:
        hdd_gb = re.findall(r'(\d+)GB\s*HDD', mem)
        hdd_size += sum([int(x) for x in hdd_gb])
        hdd_tb = re.findall(r'(\d+)TB\s*HDD', mem)
        hdd_size += sum([int(x) * 1024 for x in hdd_tb])

    # 3. Flash Memory
    has_flash = 1 if 'Flash' in mem else 0
    flash_size = 0
    if has_flash:
        # استخراج Flash به GB
        flash_gb = re.findall(r'(\d+)GB\s*Flash', mem)
        flash_size += sum([int(x) for x in flash_gb])
        # استخراج Flash به TB (تبدیل به GB)
        flash_tb = re.findall(r'(\d+)TB\s*Flash', mem)
        flash_size += sum([int(x) * 1024 for x in flash_tb])

    # 4. Hybrid (SSHD)
    has_hybrid = 1 if ('Hybrid' in mem or 'SSHD' in mem) else 0
    hybrid_size = 0
    if has_hybrid:
        # استخراج Hybrid به GB
        hybrid_gb = re.findall(r'(\d+)GB\s*(?:Hybrid|SSHD)', mem)
        hybrid_size += sum([int(x) for x in hybrid_gb])
        # استخراج Hybrid به TB (تبدیل به GB)
        hybrid_tb = re.findall(r'(\d+)TB\s*(?:Hybrid|SSHD)', mem)
        hybrid_size += sum([int(x) * 1024 for x in hybrid_tb])

    # 5. محاسبه کل حافظه (همه انواع)
    total_memory = ssd_size + hdd_size + flash_size + hybrid_size

    return has_ssd, ssd_size, has_hdd, hdd_size, has_flash, flash_size, has_hybrid, hybrid_size, total_memory

# اعمال روی دیتافریم
df['Has_SSD'], df['SSD_Size_GB'], df['Has_HDD'], df['HDD_Size_GB'], df['Has_Flash'], df['Flash_Size_GB'], df['Has_Hybrid'], df['Hybrid_Size_GB'], df['Total_Memory_GB'] = zip(*df['Memory'].apply(extract_all_memory_features))

# حذف ستون اصلی Memory
df.drop(columns=['Memory'], inplace=True)

print("✅ پردازش کامل شد!")
print(f"✅ {len(df)} رکورد پردازش شد")
```
