# **Preprocessing the " ScreenResolution " column :**


```python
import numpy as np
import re
```

```python
# ۱: IPS  ,  Touchscreen
df['IPS'] = df['ScreenResolution'].apply(lambda x: 1 if 'IPS' in x else 0)
df['Touchscreen'] = df['ScreenResolution'].apply(
lambda x: 1 if 'Touchscreen' in x else 0)
```

```python
# 2.1: Height and width extraction function
def extract_resolution(res):
  match = re.search(r'(\d+)x(\d+)', res)
  if match:
    return int(match.group(1)), int(match.group(2))
  return None, None


# 2.2: Creating columns X_res and Y_res
res_tuples = df['ScreenResolution'].apply(extract_resolution)
df['X_res'], df['Y_res'] = zip(*res_tuples)
```

```python
# 4: PPI calculation
df['PPI'] = (
(df['X_res'] ** 2 + df['Y_res'] ** 2) ** 0.5 / df['Inches']).round(2)
```

# **Preprocessing the " CPU " column :**

```python
print("تعداد تنوع متنی در ستون Cpu:", df['Cpu'].nunique())
```

```python
# ۱. Extracting frequency (speed) as a numerical variable
def extract_cpu_speed(cpu_str):
  match = re.search(r'(\d+(?:\.\d+)?)GHz', cpu_str)
  if match:
    return float(match.group(1))
  return None

df['Cpu_Speed_GHz'] = df['Cpu'].apply(extract_cpu_speed)

# ۲. Processor categories
def fetch_cpu_brand(cpu_str):
  if 'Intel Core i7' in cpu_str:
    return 'Intel Core i7'
  elif 'Intel Core i5' in cpu_str:
    return 'Intel Core i5'
  elif 'Intel Core i3' in cpu_str:
    return 'Intel Core i3'
  elif 'Celeron' in cpu_str or 'Pentium' in cpu_str:
    return 'Intel Budget'
  elif 'Intel' in cpu_str:
    return 'Intel Other'
  elif 'AMD' in cpu_str:
    return 'AMD'
  else:
    return 'Intel Other'

df['Cpu_Brand'] = df['Cpu'].apply(fetch_cpu_brand)
print(df['Cpu_Brand'].value_counts())
```

# **Preprocessing the " Nemory " column :**

```python
def extract_memory_features(memory_str):
  ssd, hdd, flash, hybrid = 0, 0, 0, 0

  # ۱. Convert TB to GB
  memory_str = memory_str.replace('1.0TB', '1024GB').replace('1TB', '1024GB')
  memory_str = memory_str.replace('2TB', '2048GB')

  # ۲. Separation, for example, 128GB SSD + 1TB HDD
  parts = memory_str.split('+')

  for part in parts:
    part = part.strip()
    match = re.search(r'(\d+)GB', part)
    if match:
      size = int(match.group(1))
      if 'SSD' in part:
        ssd += size
      elif 'HDD' in part:
        hdd += size
      elif 'Flash' in part:
        flash += size
      elif 'Hybrid' in part:
        hybrid += size

  return ssd, hdd, flash, hybrid


# Create 4 new numeric columns
(
    df['SSD_GB'],
    df['HDD_GB'],
    df['Flash_Storage_GB'],
    df['Hybrid_GB'],
) = zip(*df['Memory'].apply(extract_memory_features))

df[['Memory', 'SSD_GB', 'HDD_GB', 'Flash_Storage_GB', 'Hybrid_GB']].head(10)
```
